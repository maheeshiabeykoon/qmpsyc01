```r
# Create a vector of exam scores
scores <- c(72, 85, 91, 68, 77, 88, 95, 81, 74, 90)
print(scores)
```

## Creating a data frame
A data frame is one of the most common structures used for statistical analysis in R.

```r
students <- data.frame(
  id = 1:10,
  study_hours = c(2, 4, 5, 3, 6, 7, 4, 8, 5, 6),
  exam_score = scores
)

students
```

## Scatter plot

We can visualize the relationship between study hours and exam scores.

```r
hist(
  scores,
  main = "Distribution of Exam Scores",
  xlab = "Exam Score",
  breaks = 5
)
```

## Linear regression

We can fit a simple linear regression model predicting exam score from study hours.

```r
model <- lm(
  exam_score ~ study_hours,
  data = students
)

summary(model)
```

## Regression plot

We can add the regression line to the scatter plot.

```r
plot(
  students$study_hours,
  students$exam_score,
  main = "Study Hours and Exam Score",
  xlab = "Study Hours",
  ylab = "Exam Score",
  pch = 19
)

abline(model)
```
